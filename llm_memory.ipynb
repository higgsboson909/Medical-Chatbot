```python
# load pdfs

from langchain_community.document_loaders import PyPDFDirectoryLoader


medical_docs_dir = "/home/debo/workspace/github.com/Medical-Chatbot/med_docs/"
def load_medical_docs (dir):
    loader = PyPDFDirectoryLoader(dir, 
                                  glob="*.pdf"
    ) 
    medical_docs = loader.load()
    return medical_docs

medical_docs = load_medical_docs(medical_docs_dir)

```

```python
# create chucks of loaded data

from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50,
)

chunks = text_splitter.split_documents(medical_docs)
print(len(chunks))
```

```python
# get embedding model from hugging face 
from langchain_huggingface import HuggingFaceEmbeddings
def get_embedding_model ():
    embedding_model = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")
    return embedding_model

embedding_model = get_embedding_model()

    
```

```python
# load vector embeddings to FIASS store

from langchain_community.vectorstores import FAISS

DB_FAISS_PATH = "vectorstore/db-faiss"
db=FAISS.from_documents(chunks, embedding_model)
db.save_local(DB_FAISS_PATH)
```
