```python
# load pdfs

from langchain_community.document_loaders import PyPDFDirectoryLoader


medical_docs_dir = "/home/debo/workspace/github.com/Medical-Chatbot/med_docs/"
def load_medical_docs (dir):
    loader = PyPDFDirectoryLoader(dir, 
                                  glob="*.pdf"
    ) 
    medical_docs = loader.load()
    return medical_docs

medical_docs = load_medical_docs(medical_docs_dir)

```

```python
# create chucks of loaded data

from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50,
)

chunks = text_splitter.split_documents(medical_docs)
print(len(chunks))
```
