```python
from langchain_huggingface import HuggingFaceEndpoint
import os

HF_TOKEN = os.environ.get("HF_TOKEN")

# setup llm (mistral with hugging face)

hf_repo_id = "mistralai/Mistral-7B-Instruct-v0.3"
def load_llm(repo_id, api_token):
    llm = HuggingFaceEndpoint(
        repo_id=repo_id,
        max_length=128,
        temperature=0.3,
        huggingfacehub_api_token=api_token
    )
    return llm

MISTRAL_AI = load_llm(hf_repo_id, HF_TOKEN)
       
```

```python
from langchain_core.prompts import PromptTemplate

# create custom prompt
custom_prompt_template = f"""
You are a helpful and knowledgeable medical assistant. You will be provided with context extracted from trusted medical resources. Your task is to carefully read this context and answer the user's question based solely on the information provided. 

If the answer to the user's question is not found within the context, respond by politely stating that the information is unavailable in the provided material, and advise consulting a qualified healthcare professional.

Do not make up information, guesses, or assumptions beyond what is given in the context.

---

Context:
{context}

---

User's Question:
{question}

---

Your Answer:
"""

def set_custom_prompt(template):
    prompt=PromptTemplate(template=custom_prompt_template, input_variables=["context", "question"])
    return prompt

prompt = set_custom_prompt(custom_prompt_template)
```
